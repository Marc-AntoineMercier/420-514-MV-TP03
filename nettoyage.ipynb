### Vérifications :

```python
df_horaires.info()
```

```python
df_horaires.describe(include='all')
```

```python
df_horaires[(df_horaires['restaurant_id'] == '')]
```

```python
df_horaires.isna().sum()
```

```python
# Vérification du nombre de restaurants sans horaires (fermés?)
jours = ['lundi', 'mardi', 'mercredi', 'jeudi', 'vendredi', 'samedi', 'dimanche']
df_horaires[(df_horaires[jours] == '0:0-0:0').all(axis=1)]
```

```python
# Vérification du nombre de restaurants sans horaires (avec des NaN)
df_horaires[(df_horaires[jours].isna()).all(axis=1)]
```

### Nettoyage :

```python
# Convertir les NaN en 0:0-0:0 pour signifier une fermeture ou horaire manquant
df_horaires[jours] = df_horaires[jours].fillna('0:0-0:0')
df_horaires[(df_horaires[jours].isna()).all(axis=1)]
```

```python
# Conversion vers le types approprié
df_horaires['restaurant_id'] = df_horaires['restaurant_id'].astype('string')
df_horaires[jours] = df_horaires[jours].astype('string')
df_horaires.info()
```

```python
df_horaires.isna().sum()
```

# Création du Dataset `Business`

```python
df_business = df_restaurants[['restaurant_id', 'nom', 'ville', 'moyenne_etoiles']].rename(columns={'moyenne_etoiles': 'etoiles'})
review_count = df_avis.groupby('restaurant_id')['avis_id'].count().reset_index(name='review_count')
df_business = pd.merge(df_business, review_count, how='left', on='restaurant_id')
df_business = pd.merge(df_business, df_categories.groupby('restaurant_id')['categorie'].apply(lambda x: ', '.join(x.unique())).reset_index(), how='left', on='restaurant_id')
df_business = df_business.rename(columns={'restaurant_id': 'business_id'})
df_business
```
