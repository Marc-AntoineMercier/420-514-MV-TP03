## Création du Dataset `Business`

```python
df_business = df_restaurants[['restaurant_id', 'nom', 'ville', 'moyenne_etoiles']].rename(columns={'moyenne_etoiles': 'etoiles'})

review_count = df_avis.groupby('restaurant_id')['avis_id'].count().reset_index(name='review_count')

df_business = pd.merge(df_business, review_count, how='left', on='restaurant_id')
df_business = pd.merge(df_business, df_categories.groupby('restaurant_id')['categorie'].apply(lambda x: ', '.join(x.unique())).reset_index(), how='left', on='restaurant_id')
df_business = df_business.rename(columns={'restaurant_id': 'business_id', 'categorie': 'categories'})
df_business['categories'] = df_business['categories'].astype('string')
df_business
```
