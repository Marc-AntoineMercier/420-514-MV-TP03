# Nettoyage des données

## Avis

### Vérifications :

```python
# Vérification des colonnes
df_avis
```

```python
# Vérification des types
df_avis.info()
```

```python
# Vérification rapide
df_avis.describe(include="all")

# Tous des ID d'avis uniques
```

```python
# Vérification des données manquantes
df_avis.isna().sum()
```

```python
# Vérification si le nombre d'étoiles est bien entre 1 et 5
df_avis[(df_avis['etoiles'] < 1) & (df_avis['etoiles'] > 5)]
```

```python
# Verification s'il existe des nombres négatifs
df_avis[(df_avis['useful'] < 0) & (df_avis['funny'] < 0) & (df_avis['cool'] < 0)]
```

### Nettoyages :

```python
# Conversion vers les types appropriés
df_avis["avis_id"] = df_avis["avis_id"].astype("string")
df_avis["restaurant_id"] = df_avis["restaurant_id"].astype("string")
df_avis["utilisateur_id"] = df_avis["utilisateur_id"].astype("string")
df_avis["etoiles"] = df_avis["etoiles"].astype("int64")
df_avis['date'] = pd.to_datetime(df_avis['date'], errors='coerce')
df_avis.info()
```

```python
# Vérification s'il n'y a pas de dates invalides
df_avis['date'].isna().sum()
```

```python
# Vérification des types
df_utilisateurs.info()
```

```python
# Vérification rapide
df_utilisateurs.describe(include="all")

# Tous des ID Uniques
# Minimum de 0 et maximum de 1 pour elite
# Aucun nombre négatif
```
