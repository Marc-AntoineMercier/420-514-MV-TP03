## Préparer review count

```python
review_count_join = (df_avis
    .groupby("restaurant_id")
    .size()
    .reset_index(name = "review_count_total"))
```

```python
positive_reviews = (df_avis[df_avis["etoiles"] >= 4]
    .groupby("restaurant_id")
    .size()
    .reset_index(name = "review_positive"))
```

```python
review = merge(review_count_join, 
               positive_reviews, 
               left_on= "restaurant_id", 
               right_on="restaurant_id", 
               how="inner")
```

```python
data = data.merge(review, 
             left_on  = "restaurant_id", 
             right_on = "restaurant_id", 
             how      = "inner")
```

```python
data["positive_ratio"] = floor((data["review_positive"] / data["review_count_total"]) * 100) / 100
```

## Préparations checkin total

```python
checkins_total = (df_checkin.groupby("restaurant_id")
    .size()
    .reset_index(name="checkins_total"))
```

```python
data = data.merge(checkins_total, 
             left_on = "restaurant_id", 
             right_on = "restaurant_id", 
             how = "inner")
```

```python
name_counts = data["nom"].value_counts()
```

```python
data["is_chain"] = data["nom"].map(lambda x: name_counts[x] >= 3)
```

```python
prix_moyen = (df_services
    .groupby("restaurant_id")["prix"]
    .mean()
    .reset_index(name="prix_moyen"))
```

```python
data = data.merge(prix_moyen, 
                  left_on = "restaurant_id", 
                  right_on = "restaurant_id", 
                  how = "left")
```

# Graphiques

## Les catégories les plus appréciées par villes

### Préparation des données

```python
df_categories_aimees = df_categories.merge(df_restaurants[['restaurant_id', 'ville', 'moyenne_etoiles']], how='left', on='restaurant_id')
df_categories_aimees = df_categories_aimees.merge(review_count[['restaurant_id', 'review_count']], how='left', on='restaurant_id' )

df_categories_aimees = (df_categories_aimees.groupby(['ville', 'categorie']).agg(
    etoiles=('moyenne_etoiles', 'mean'), 
    avis=('review_count', 'sum'))).reset_index()

liste_villes = ['Montreal', 'Laval', 'Toronto', 'Longueuil', 'Brampton', 'Etobicoke', 'Markham', 'Calgary', 'Mississauga', 'Saint-Leonard']

# df_categories_aimees[df_categories_aimees['ville'] == 'Toronto'].head(30)
```

### Normalisation des données 

```python
avis_ville_moyenne = df_categories_aimees.groupby('ville')['avis'].transform('mean')
df_categories_aimees = df_categories_aimees[df_categories_aimees['avis'] >= avis_ville_moyenne]

df_top3_par_villes = (df_categories_aimees.sort_values(['ville', 'etoiles'], ascending=[True, False]).groupby('ville').head(3))
```

```python
df_top3_par_villes[df_top3_par_villes['ville'] == 'Montreal']
```

### Préparation du graphique

```python
villes = df_top3_par_villes[df_top3_par_villes['ville'].isin(liste_villes)]

fig, axes = plt.subplots(
    nrows=len(liste_villes), 
    ncols=1, 
    figsize=(10, 3 * len(liste_villes)),
    constrained_layout=True
)

for ax, ville in zip(axes, liste_villes):
    
    subset = villes[villes['ville'] == ville]
    subset = subset.sort_values('etoiles')

    ax.barh(subset['categorie'], subset['etoiles'])
    ax.set_title(f'\n\n{ville}', fontsize=14)
    ax.xaxis.set_major_formatter(plt.FormatStrFormatter('%.2f'))

    min_etoiles = subset['etoiles'].min() - 0.05
    max_etoiles = subset['etoiles'].max() + 0.05
    ax.set_xlim(min_etoiles, max_etoiles)

    for i, (etoiles, avis) in enumerate(zip(subset['etoiles'], subset['avis'])):
        ax.text(etoiles + (max_etoiles - min_etoiles) * 0.01, i, f'{etoiles:.2f}★', va='center')

fig.suptitle('Top 3 des types de commerces les plus aimées (par ville)', fontsize=20, y=1.02)

plt.show()
```
